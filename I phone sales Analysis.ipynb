```python
import pandas as pd 
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

```python
data = pd.read_csv("Apple_product.csv")           
```

```python
print(data.isnull().sum())     # first step to check the null values 
```

```python
data.describe()                 # statistical  value of numeric data only
```

```python
high_rating = data.sort_values(by = ["Star Rating"] , ascending =  False)
```

```python
iphones = high_rating["Product Name"].value_counts()
labels = iphones.index
counts= high_rating["Number Of Ratings"]


figure = px.bar(high_rating , x=lables,y=counts,
                title ="Numbers of ratings of highest rated i phones ")
figure.show()
```

```python
iphones = high_rating["Product Name"].value_counts()
labels = iphones.index
counts= high_rating["Number Of Reviews"]


figure = px.bar(high_rating , x=lables,y=counts,
                title ="Numbers of Reviews of highest rated i phones ")
figure.show()
```
